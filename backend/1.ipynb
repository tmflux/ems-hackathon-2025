```python
def load_mendeley_month(path):
    xls = pd.ExcelFile(path)

    # Case 1: 2021–2023 monthly files (SCADA style)
    if "Sheet1" in xls.sheet_names:
        df = pd.read_excel(path, sheet_name="Sheet1", header=1)

        if "Unnamed: 0" in df.columns:
            df = df.rename(columns={"Unnamed: 0": "time"})
        df["time"] = pd.to_datetime(df["time"])
        df = df.set_index("time").sort_index()

        rename_map = {
            "NLDC_DEMAND|P": "demand_MW",
            "ALL_IND_SOLAR|P": "solar_MW",
            "ALL_INDIA_WIND|P": "wind_MW",
            "Solar+Wind": "solar_wind_MW",
            "Solar": "solar_MW_cleaned"
        }
        df = df.rename(columns=rename_map)

        out = pd.DataFrame(index=df.index)
        if "demand_MW" in df.columns:
            out["demand_kW"] = df["demand_MW"].astype(float) * 1000.0
        if "solar_wind_MW" in df.columns:
            out["solar_wind_kW"] = df["solar_wind_MW"].astype(float) * 1000.0
        if "solar_MW" in df.columns:
            out["solar_kW"] = df["solar_MW"].astype(float) * 1000.0
        if "wind_MW" in df.columns:
            out["wind_kW"] = df["wind_MW"].astype(float) * 1000.0
        if "solar_MW_cleaned" in df.columns:
            out["solar_cleaned_kW"] = df["solar_MW_cleaned"].astype(float) * 1000.0

    # Case 2: 2024–2025 file (Report sheet, clean headers)
    elif "Report" in xls.sheet_names:
        df = pd.read_excel(path, sheet_name="Report", header=0)

        df["Timestamp"] = pd.to_datetime(df["Timestamp"], dayfirst=True)
        df = df.set_index("Timestamp").sort_index()

        out = pd.DataFrame(index=df.index)
        if "Demand (MW)" in df.columns:
            out["demand_kW"] = df["Demand (MW)"].astype(float) * 1000.0
        if "Solar (MW)" in df.columns:
            out["solar_kW"] = df["Solar (MW)"].astype(float) * 1000.0
        if "Wind (MW)" in df.columns:
            out["wind_kW"] = df["Wind (MW)"].astype(float) * 1000.0
        if "Total Generation (MW)" in df.columns:
            out["total_gen_kW"] = df["Total Generation (MW)"].astype(float) * 1000.0

    else:
        raise ValueError(f"No recognized sheet found in {path}: {xls.sheet_names}")

    # Resample to 5s
    out = out.resample("5s").interpolate()
    return out
df = load_mendeley_month(r"D:\ems-hackathon-2025\backend\data\Mendeley\Electricity Demand, Solar and Wind Generation Data\January 2024- June 2025.xlsx")
print(df.head())
```

```python
import pandas as pd
import numpy as np
import glob
import os

# -----------------------------
# Kaggle loaders
# -----------------------------
def load_kaggle_generation(path):
    df = pd.read_csv(path, parse_dates=["DATE_TIME"], dayfirst=True)
    df = df.sort_values("DATE_TIME")
    agg = df.groupby("DATE_TIME").agg({
        "DC_POWER": "sum",
        "AC_POWER": "sum"
    }).reset_index()
    # 1-minute resolution for memory efficiency
    agg = agg.set_index("DATE_TIME").resample("1min").interpolate()
    return agg.rename(columns={"DC_POWER": "dc_kW", "AC_POWER": "ac_kW"})


def load_kaggle_weather(path):
    df = pd.read_csv(path, parse_dates=["DATE_TIME"], dayfirst=True)
    df = df.sort_values("DATE_TIME")
    agg = df.groupby("DATE_TIME").agg({
        "AMBIENT_TEMPERATURE": "mean",
        "MODULE_TEMPERATURE": "mean",
        "IRRADIATION": "mean"
    }).reset_index()
    agg = agg.set_index("DATE_TIME").resample("1min").interpolate()
    return agg.rename(columns={
        "AMBIENT_TEMPERATURE": "ambient_C",
        "MODULE_TEMPERATURE": "module_C",
        "IRRADIATION": "irradiance"
    })

# -----------------------------
# Mendeley loaders
# -----------------------------
def load_mendeley_month(path):
    xls = pd.ExcelFile(path)

    if "Sheet1" in xls.sheet_names:
        df = pd.read_excel(path, sheet_name="Sheet1", header=1)
        if "Unnamed: 0" in df.columns:
            df = df.rename(columns={"Unnamed: 0": "time"})
        df["time"] = pd.to_datetime(df["time"])
        df = df.set_index("time").sort_index()
        rename_map = {
            "NLDC_DEMAND|P": "demand_MW",
            "ALL_IND_SOLAR|P": "solar_MW",
            "ALL_INDIA_WIND|P": "wind_MW",
            "Solar+Wind": "solar_wind_MW",
            "Solar": "solar_MW_cleaned"
        }
        df = df.rename(columns=rename_map)
        out = pd.DataFrame(index=df.index)
        if "demand_MW" in df.columns:
            out["demand_kW"] = df["demand_MW"].astype(float) * 1000.0
        if "solar_wind_MW" in df.columns:
            out["solar_wind_kW"] = df["solar_wind_MW"].astype(float) * 1000.0
        if "solar_MW" in df.columns:
            out["solar_kW"] = df["solar_MW"].astype(float) * 1000.0
        if "wind_MW" in df.columns:
            out["wind_kW"] = df["wind_MW"].astype(float) * 1000.0
        if "solar_MW_cleaned" in df.columns:
            out["solar_cleaned_kW"] = df["solar_MW_cleaned"].astype(float) * 1000.0

    elif "Report" in xls.sheet_names:
        df = pd.read_excel(path, sheet_name="Report", header=0)
        df["Timestamp"] = pd.to_datetime(df["Timestamp"], dayfirst=True, errors="coerce")
        df = df.dropna(subset=["Timestamp"])
        df = df.set_index("Timestamp").sort_index()
        out = pd.DataFrame(index=df.index)
        if "Demand (MW)" in df.columns:
            out["demand_kW"] = df["Demand (MW)"].astype(float) * 1000.0
        if "Solar (MW)" in df.columns:
            out["solar_kW"] = df["Solar (MW)"].astype(float) * 1000.0
        if "Wind (MW)" in df.columns:
            out["wind_kW"] = df["Wind (MW)"].astype(float) * 1000.0
        if "Total Generation (MW)" in df.columns:
            out["total_gen_kW"] = df["Total Generation (MW)"].astype(float) * 1000.0

    else:
        raise ValueError(f"No recognized sheet found in {path}: {xls.sheet_names}")

    # Resample to 1min for memory efficiency
    out = out.resample("1min").interpolate()
    return out


def load_mendeley_all(folder_path, year=None):
    all_files = glob.glob(os.path.join(folder_path, "*.xlsx"))
    dfs = []
    for f in sorted(all_files):
        try:
            df = load_mendeley_month(f)
            if year:
                df = df[df.index.year == int(year)]
            if not df.empty:
                dfs.append(df)
        except Exception as e:
            print(f"Skipping {f}: {e}")
    return pd.concat(dfs).sort_index() if dfs else pd.DataFrame()

# -----------------------------
# Alignment (Option B: relative rescaling)
# -----------------------------
def resample_align_relative(pv_df, demand_df, weather_df=None):
    pv_df = pv_df.reset_index(drop=True)
    demand_df = demand_df.reset_index(drop=True)
    if weather_df is not None:
        weather_df = weather_df.reset_index(drop=True)

    min_len = min(len(pv_df), len(demand_df), *( [len(weather_df)] if weather_df is not None else [] ))
    if min_len == 0:
        raise ValueError(f"Empty dataset: PV={len(pv_df)}, Demand={len(demand_df)}")

    pv_df = pv_df.iloc[:min_len]
    demand_df = demand_df.iloc[:min_len]
    if weather_df is not None:
        weather_df = weather_df.iloc[:min_len]

    idx = pd.timedelta_range(start="0s", periods=min_len, freq="1min")
    df = pd.DataFrame(index=idx).join(pv_df).join(demand_df)
    if weather_df is not None:
        df = df.join(weather_df)

    numeric_cols = df.select_dtypes(include=[np.number]).columns
    df[numeric_cols] = df[numeric_cols].interpolate()
    df = df.dropna()

    seconds = df.index / np.timedelta64(1, "s")
    hours = (seconds // 3600) % 24
    price = np.where((hours >= 18) & (hours <= 22), 7.0, 3.0)

    pv_series = df["ac_kW"] if "ac_kW" in df.columns else df["dc_kW"]
    return pv_series.values, df["demand_kW"].values, price
```

```python
import pandas as pd
import glob
import os

# --- Kaggle PV loader ---
def load_kaggle_generation_demo(path):
    df = pd.read_csv(path, parse_dates=["DATE_TIME"], dayfirst=True)
    df = df[["DATE_TIME", "AC_POWER", "DC_POWER"]].sort_values("DATE_TIME")
    print("Kaggle PV Data Sample:")
    print(df.head())
    return df

# --- Robust Mendeley loader ---
def load_mendeley_month(path):
    xls = pd.ExcelFile(path)
    out = None

    # Sheet1
    if "Sheet1" in xls.sheet_names:
        df = pd.read_excel(path, sheet_name="Sheet1", header=1)
        if "Unnamed: 0" in df.columns:
            df = df.rename(columns={"Unnamed: 0": "time"})
        df["time"] = pd.to_datetime(df["time"])
        rename_map = {"Solar+Wind": "solar_wind_kW"}
        df = df.rename(columns=rename_map)
        out = df[["time", "solar_wind_kW"]]
    
    # Report
    elif "Report" in xls.sheet_names:
        df = pd.read_excel(path, sheet_name="Report", header=0)
        df["Timestamp"] = pd.to_datetime(df["Timestamp"], dayfirst=True, errors="coerce")
        df = df.dropna(subset=["Timestamp"])
        rename_map = {"Total Generation (MW)": "solar_wind_kW"}
        df = df.rename(columns=rename_map)
        out = df[["Timestamp", "solar_wind_kW"]]
        out = out.rename(columns={"Timestamp": "time"})

    if out is not None:
        out.loc[:, "solar_wind_kW"] = out["solar_wind_kW"].astype(float) * 1000
        out = out.sort_values("time")
    return out

def load_mendeley_demo(folder_path):
    all_files = glob.glob(os.path.join(folder_path, "*.xlsx"))
    df_list = []
    for f in sorted(all_files):
        try:
            df = load_mendeley_month(f)
            if df is not None and not df.empty:
                df_list.append(df)
        except Exception as e:
            print(f"Skipping {f}: {e}")
    if df_list:
        df_all = pd.concat(df_list, ignore_index=True).sort_values("time")
        print("\nMendeley Load Data Sample:")
        print(df_all.head())
        return df_all
    else:
        print("No valid Excel files found.")
        return pd.DataFrame()

# ---------------- DEMO USAGE -----------------
pv_demo_path = r"D:\ems-hackathon-2025\backend\data\Kaggle\Plant_1_Generation_Data.csv"
load_demo_folder  = r"D:\ems-hackathon-2025\backend\data\Mendeley\Electricity Demand, Solar and Wind Generation Data"


pv_df = load_kaggle_generation_demo(pv_demo_path)
load_df = load_mendeley_demo(load_demo_folder)
```

```python
import json
import time
import paho.mqtt.client as mqtt

MQTT_HOST = "localhost"
MQTT_PORT = 1883

client = mqtt.Client()
client.connect(MQTT_HOST, MQTT_PORT)

actions = [1, 2, 3]  # 1=Charge Battery, 2=Discharge Battery, 3=Charge EV

while True:
    for action in actions:
        payload = {"action": action}
        client.publish("ems/site1/actions", json.dumps(payload))
        print(f"Sent RL action: {action}")
        time.sleep(10)  # wait 10 seconds before sending next action
```
